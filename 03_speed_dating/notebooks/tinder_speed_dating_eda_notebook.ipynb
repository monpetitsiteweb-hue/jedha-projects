# Tinder Speed Dating Analysis

## 1. Objective

The goal of this project is to understand what makes people interested in each other and willing to go on a second date.

The dataset comes from a speed dating experiment in which participants rated each other on several attributes such as attractiveness, sincerity, intelligence, fun, ambition, and shared interests.

Each row represents one speed date seen by one participant, so a date between two people gives two rows.

The next cell imports the libraries and loads the file. The output shows 8,378 rows and 195 columns.

## 8b. Is the effect of date order real? (inferential check)

The rates above come from a sample, so part of any difference could be chance. A 95% confidence interval gives the range of plausible values for the difference between two rates. If the interval does not contain 0, chance alone is an unlikely explanation.

This cell computes that interval for the first date and for the last date. The output shows +8.5 points for the first date, from +4.2 to +12.8, and +3.4 points for the last date, from -0.9 to +7.7.

```python
def diff_rate_ci(group_a, group_b):
    p_a, p_b = group_a.mean(), group_b.mean()
    se = np.sqrt(p_a * (1 - p_a) / len(group_a) + p_b * (1 - p_b) / len(group_b))
    diff = p_a - p_b
    return {"rate": p_a, "rate_others": p_b, "n": len(group_a), "diff": diff,
            "ci_low": diff - 1.96 * se, "ci_high": diff + 1.96 * se}

first = df_order["is_first"] == 1
last = df_order["is_last"] == 1
order_ci = pd.DataFrame({
    "first date vs others": diff_rate_ci(df_order.loc[first, "dec"], df_order.loc[~first, "dec"]),
    "last date vs others": diff_rate_ci(df_order.loc[last, "dec"], df_order.loc[~last, "dec"]),
}).T
order_ci.round(3)
```

The interval above treats all rows as independent, but the decisions of one person are not. This cell checks it with a bootstrap by participant: it draws 551 participants with replacement, keeps all the rows of each one, and recomputes both differences. It repeats this 2,000 times with a fixed seed, so the result does not change between runs. The 2.5th and 97.5th percentiles of these differences give the 95% interval.


```python
rng = np.random.default_rng(42)
rows_by_person = {
    iid: (person["dec"].to_numpy(), person["is_first"].to_numpy() == 1, person["is_last"].to_numpy() == 1)
    for iid, person in df_order.groupby("iid")
}
participants = list(rows_by_person)

boot_first, boot_last = [], []
for _ in range(2000):
    sample = rng.choice(participants, len(participants))
    dec = np.concatenate([rows_by_person[iid][0] for iid in sample])
    is_first = np.concatenate([rows_by_person[iid][1] for iid in sample])
    is_last = np.concatenate([rows_by_person[iid][2] for iid in sample])
    boot_first.append(dec[is_first].mean() - dec[~is_first].mean())
    boot_last.append(dec[is_last].mean() - dec[~is_last].mean())

bootstrap_ci = pd.DataFrame(
    [np.percentile(boot_first, [2.5, 97.5]), np.percentile(boot_last, [2.5, 97.5])],
    index=["first date vs others", "last date vs others"],
    columns=["ci_low", "ci_high"],
)
bootstrap_ci.round(3)
```

### Interpretation

Being the first date of the night is associated with a yes rate 8.5 points higher (50% vs 41%), and the 95% interval, from +4.2 to +12.8 points, does not contain 0. Being the last date shows no clear effect: the interval, from -0.9 to +7.7 points, contains 0. Being first goes with a higher yes rate, and the data do not show an advantage for being last.

Two limits. The same person rates several partners, so the rows are not fully independent. The bootstrap by participant takes this into account and gives almost the same result: from +4.3 to +12.5 points for the first date, and from -0.5 to +7.0 for the last. And this shows an association, not a cause: the data do not tell us why first dates do better.

## 9. Stated Preferences vs Revealed Importance

We compare what participants say they value before the event (`attr1_1`, ...) with what their decisions reveal: the correlation, computed in section 6, between the rating given to the partner on each attribute and the decision `dec`. Follow-up answers such as `attr7_2` are not used, because they are still statements, not decisions.

The two measures have different scales, so we compare their ranks. Stated preferences exclude waves 6 to 9, because the data key describes another scale for them, from 1 to 10. Including them does not change the ranks.

The output shows attractiveness first in both overall (23.98 points, correlation 0.487). By gender, women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both. Fun goes from 3rd to 2nd and shared interests from 5th to 3rd, while intelligence falls from 2nd to 4th and sincerity from 4th to 5th. Ambition is last in both.

```python
mask_same_scale = ~df["wave"].between(6, 9)

# 1. Stated preferences (survey before the event, 100 points to share)
stated_preferences = (
    df.loc[mask_same_scale, ["attr1_1", "sinc1_1", "intel1_1", "fun1_1", "amb1_1", "shar1_1"]]
    .mean()
    .reset_index()
)
stated_preferences["attribute"] = stated_preferences["index"].str.replace("1_1", "").str.replace("_", "")
stated_preferences = stated_preferences[["attribute", 0]]
stated_preferences.columns = ["attribute", "stated_importance"]

# 2. Revealed importance: correlation between each rating given to the partner and the decision (section 6)
revealed = (
    corr_with_decision
    .rename(columns={"variable": "attribute", "dec": "correlation_with_dec"})
    .query("attribute in @stated_preferences.attribute")
)

# 3. Merge and rank
comparison_pref = stated_preferences.merge(revealed, on="attribute")
comparison_pref["stated_rank"] = comparison_pref["stated_importance"].rank(ascending=False).astype(int)
comparison_pref["revealed_rank"] = comparison_pref["correlation_with_dec"].rank(ascending=False).astype(int)
comparison_pref = comparison_pref.sort_values("revealed_rank").reset_index(drop=True)

comparison_pref
```

This chart draws one line per attribute, from its stated rank to its revealed rank. A line going up means that the attribute matters more in decisions than people say. It shows the rise of shared interests and fun, and the fall of intelligence and sincerity.

```python
fig = go.Figure()
for _, row in comparison_pref.iterrows():
    fig.add_scatter(x=["Stated", "Revealed by decisions"], y=[row["stated_rank"], row["revealed_rank"]],
                    mode="lines+markers", name=row["attribute"])
fig.update_yaxes(autorange="reversed", dtick=1, title="rank (1 = most important)")
fig.update_layout(title="Stated rank vs rank revealed by decisions")
fig.show()
```

This cell checks the choice about waves 6 to 9. The data key describes a scale from 1 to 10 for them, but their answers already total about 100 in the file. Including them gives the same ranks, so the choice does not change the result.

```python
stated_cols = ["attr1_1", "sinc1_1", "intel1_1", "fun1_1", "amb1_1", "shar1_1"]
print("Total points, waves 6 to 9:", df.loc[df["wave"].between(6, 9), stated_cols].mean().sum().round(1))
same_ranks = df[stated_cols].mean().rank().equals(df.loc[mask_same_scale, stated_cols].mean().rank())
print("Same ranks with all the waves:", same_ranks)
```

### Interpretation

Overall, attractiveness ranks first both in what people say and in what their decisions reveal. By gender, women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both. Other shifts: shared interests move from 5th to 3rd and fun from 3rd to 2nd, while intelligence falls from 2nd to 4th and sincerity from 4th to 5th. Ambition is last in both.

These are correlations measured right after each date. They do not prove a cause: someone who already wants a second date may also rate that partner higher on everything.

## 10. Conclusion

This analysis highlights the key drivers behind second-date decisions.

### Main Drivers of Attraction

Second-date decisions are strongly linked to:

- attractiveness
- fun
- shared interests
- overall liking
- perceived reciprocity

Among these, attractiveness plays a particularly strong role in real decisions.

---

### Stated Preferences vs Actual Behavior

What participants say they value matches their decisions only in part.

- **Overall, attractiveness ranks first in both**: women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both
- **Intelligence and sincerity rank lower in decisions** (intelligence: 2nd stated, 4th revealed; sincerity: 4th stated, 5th revealed)
- **Shared interests and fun rank higher in decisions** (shared interests: 5th stated, 3rd revealed; fun: 3rd stated, 2nd revealed)

This is an association, measured with ratings given right after each date. It does not prove that people misjudge their own preferences.

---

### Answers to Key Questions

**1. What are the least desirable attributes? Does it differ by gender?**

In decisions, the attribute least related to saying yes is ambition for a male partner (correlation 0.175) and sincerity for a female partner (0.191). The top three are the same for both: attractiveness, fun, shared interests. The bottom attributes are close to each other, so this difference is small.

---

**2. How important is attractiveness vs its real impact?**

Overall, attractiveness **ranks first in both stated preferences and decisions** (correlation with the decision 0.487). By gender, women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both.

---

**3. Are shared interests more important than same racial background?**

Yes. Shared interests have a stronger relationship with second-date decisions than same-race status (correlation 0.401 vs 0.023).
This comparison is indicative and not on equal terms: the shared-interests rating is given after the date, while racial background is known before it.

---

**4. Does the order of the date matter?**

Being first is associated with more yes answers: the yes rate is 50% against 41% for other dates (difference +8.5 points, 95% CI +4.2 to +12.8). Being last shows no clear effect (95% CI -0.9 to +7.7). The data do not tell us why.

---

### Final Takeaway (Business Insight)

- **Attractiveness is the strongest of the six rated attributes (0.487), even though women state intelligence first**; overall liking (`like`, 0.513) is a broader measure
- **Compatibility weighs more in decisions than in statements**: shared interests and fun rank higher in decisions than in stated preferences, intelligence and sincerity lower

From a product perspective, Tinder could improve match quality by:

- emphasizing shared interests and compatibility signals
- helping users better understand mutual attraction
- not relying solely on visual profiles
