# Tinder Speed Dating Analysis

## 1. Objective

What makes people want to see each other again? That is the question of this project.

The data comes from a speed dating experiment. After each four-minute date, participants said whether they wanted a second date, and rated their partner on six attributes: attractiveness, sincerity, intelligence, fun, ambition and shared interests.

One row is one date, seen by one participant. A date between two people gives two rows.

The next cell imports the libraries and loads the file. The output shows 8,378 rows and 195 columns.

## 8b. Is the effect of date order real? (inferential check)

The rates above come from a sample, so part of any difference could be chance. A 95% confidence interval gives the range of plausible values for the difference between two rates. If the interval does not contain 0, chance alone is an unlikely explanation.

This cell computes that interval for the first date and for the last date. First date: +8.5 points, from +4.2 to +12.8. Last date: +3.4 points, from -0.9 to +7.7.

```python
def diff_rate_ci(group_a, group_b):
    p_a, p_b = group_a.mean(), group_b.mean()
    se = np.sqrt(p_a * (1 - p_a) / len(group_a) + p_b * (1 - p_b) / len(group_b))
    diff = p_a - p_b
    return {"rate": p_a, "rate_others": p_b, "n": len(group_a), "diff": diff,
            "ci_low": diff - 1.96 * se, "ci_high": diff + 1.96 * se}

first = df_order["is_first"] == 1
last = df_order["is_last"] == 1
order_ci = pd.DataFrame({
    "first date vs others": diff_rate_ci(df_order.loc[first, "dec"], df_order.loc[~first, "dec"]),
    "last date vs others": diff_rate_ci(df_order.loc[last, "dec"], df_order.loc[~last, "dec"]),
}).T
order_ci.round(3)
```

The interval above treats all rows as independent, but the decisions of one person are not. This cell checks it with a bootstrap by participant: it draws 551 participants with replacement, keeps all the rows of each one, and recomputes both differences. 2,000 times, with a fixed seed, so the result does not change between runs. The 2.5th and 97.5th percentiles of these differences give the 95% interval.

```python
rng = np.random.default_rng(42)
rows_by_person = {
    iid: (person["dec"].to_numpy(), person["is_first"].to_numpy() == 1, person["is_last"].to_numpy() == 1)
    for iid, person in df_order.groupby("iid")
}
participants = list(rows_by_person)

boot_first, boot_last = [], []
for _ in range(2000):
    sample = rng.choice(participants, len(participants))
    dec = np.concatenate([rows_by_person[iid][0] for iid in sample])
    is_first = np.concatenate([rows_by_person[iid][1] for iid in sample])
    is_last = np.concatenate([rows_by_person[iid][2] for iid in sample])
    boot_first.append(dec[is_first].mean() - dec[~is_first].mean())
    boot_last.append(dec[is_last].mean() - dec[~is_last].mean())

bootstrap_ci = pd.DataFrame(
    [np.percentile(boot_first, [2.5, 97.5]), np.percentile(boot_last, [2.5, 97.5])],
    index=["first date vs others", "last date vs others"],
    columns=["ci_low", "ci_high"],
)
bootstrap_ci.round(3)
```

### Interpretation

Being the first date of the night goes with a yes rate 8.5 points higher (49.9% against 41.4%), and the 95% interval, from +4.2 to +12.8 points, does not contain 0. Being the last date shows no clear effect: the interval, from -0.9 to +7.7 points, contains 0.

Two limits. The same person rates several partners, so the rows are not fully independent. The bootstrap by participant takes this into account and gives almost the same result: from +4.3 to +12.5 points for the first date, from -0.5 to +7.0 for the last. And this is a link, not a cause: the data do not tell us why first dates do better.

## 9. Stated Preferences vs Revealed Importance

What participants say they value before the event (`attr1_1`, ...: 100 points to share between the six attributes), against what their decisions reveal: the correlation, computed in section 6, between the rating given on each attribute and the decision `dec`. Follow-up answers such as `attr7_2` are not used: they are still statements, not decisions.

The two measures have different scales, so we compare their ranks. Stated preferences exclude waves 6 to 9, because the data key describes another scale for them, from 1 to 10. Including them does not change the ranks (checked below).

The output shows attractiveness first in both (23.98 points; correlation 0.487). By gender, women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both. Fun goes from 3rd to 2nd and shared interests from 5th to 3rd; intelligence falls from 2nd to 4th and sincerity from 4th to 5th. Ambition is last in both.

```python
mask_same_scale = ~df["wave"].between(6, 9)

# 1. Stated preferences (survey before the event, 100 points to share)
stated_preferences = (
    df.loc[mask_same_scale, ["attr1_1", "sinc1_1", "intel1_1", "fun1_1", "amb1_1", "shar1_1"]]
    .mean()
    .reset_index()
)
stated_preferences["attribute"] = stated_preferences["index"].str.replace("1_1", "").str.replace("_", "")
stated_preferences = stated_preferences[["attribute", 0]]
stated_preferences.columns = ["attribute", "stated_importance"]

# 2. Revealed importance: correlation between each rating given to the partner and the decision (section 6)
revealed = (
    corr_with_decision
    .rename(columns={"variable": "attribute", "dec": "correlation_with_dec"})
    .query("attribute in @stated_preferences.attribute")
)

# 3. Merge and rank
comparison_pref = stated_preferences.merge(revealed, on="attribute")
comparison_pref["stated_rank"] = comparison_pref["stated_importance"].rank(ascending=False).astype(int)
comparison_pref["revealed_rank"] = comparison_pref["correlation_with_dec"].rank(ascending=False).astype(int)
comparison_pref = comparison_pref.sort_values("revealed_rank").reset_index(drop=True)

comparison_pref
```

One line per attribute, from its stated rank to its rank in decisions. A line going up means the attribute matters more in decisions than people say: shared interests and fun go up, intelligence and sincerity go down.

```python
fig = go.Figure()
for _, row in comparison_pref.iterrows():
    fig.add_scatter(x=["Stated", "Revealed by decisions"], y=[row["stated_rank"], row["revealed_rank"]],
                    mode="lines+markers", name=row["attribute"])
fig.update_yaxes(autorange="reversed", dtick=1, title="rank (1 = most important)")
fig.update_layout(title="Stated rank vs rank revealed by decisions")
fig.show()
```

Check of the choice about waves 6 to 9. The data key describes a scale from 1 to 10 for them, but their answers already total about 100 in the file. Including them gives the same ranks: the choice does not change the result.

```python
stated_cols = ["attr1_1", "sinc1_1", "intel1_1", "fun1_1", "amb1_1", "shar1_1"]
print("Total points, waves 6 to 9:", df.loc[df["wave"].between(6, 9), stated_cols].mean().sum().round(1))
same_ranks = df[stated_cols].mean().rank().equals(df.loc[mask_same_scale, stated_cols].mean().rank())
print("Same ranks with all the waves:", same_ranks)
```

### Interpretation

Overall, attractiveness ranks first both in what people say and in what their decisions reveal. By gender, women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both. Shared interests move from 5th to 3rd and fun from 3rd to 2nd; intelligence falls from 2nd to 4th and sincerity from 4th to 5th. Ambition is last in both.

These are correlations measured right after each date. They do not prove a cause: someone who already wants a second date may also rate that partner higher on everything.

The same correlations, separately for each gender, for the question of the brief about male and female partners. `gender` = 0 means a woman is rating, so the partner is a man. Same top three for both: attractiveness (0.445 and 0.515), fun (0.417 and 0.407), shared interests (0.413 and 0.387). The attribute least linked to a yes: ambition for a male partner (0.175), sincerity for a female partner (0.191). The bottom values are close to each other.

```python
attributes = comparison_pref["attribute"].tolist()
corr_by_rater = pd.DataFrame({
    "partner is male (women rating)": df.loc[df["gender"] == 0, attributes + ["dec"]].corr()["dec"],
    "partner is female (men rating)": df.loc[df["gender"] == 1, attributes + ["dec"]].corr()["dec"],
}).drop("dec")
corr_by_rater.round(3)
```

The yes rate by attractiveness rating, separately for women and men, as in section 6. `gender` = 0 means a woman is rating. The yes rate rises with the rating for both: from 10.2% to 70.1% for women, from 4.8% to 80.7% for men.

```python
yes_by_attr_gender = (
    df.groupby([attr_band, df["gender"]], observed=True)["dec"]
    .mean()
    .unstack()
    .rename(columns={0: "women rating (partner is male)", 1: "men rating (partner is female)"})
)
(yes_by_attr_gender * 100).round(1)
```

For women only: the yes rate by intelligence rating, next to the yes rate by attractiveness rating. Women state intelligence first before the event. A top intelligence rating (7.5 or more) goes with 45.7% of yes, against 70.1% for a top attractiveness rating. Their decisions follow attractiveness more than intelligence.

```python
bands = {"bins": [-1, 4, 6, 7, 10], "labels": ["4 or less", "4.5 to 6", "6.5 to 7", "7.5 or more"]}
women = df[df["gender"] == 0]
women_yes = pd.DataFrame({
    "yes rate by intelligence rating": women.groupby(pd.cut(women["intel"], **bands), observed=True)["dec"].mean(),
    "yes rate by attractiveness rating": women.groupby(pd.cut(women["attr"], **bands), observed=True)["dec"].mean(),
})
(women_yes * 100).round(1)
```

## 10. Conclusion

### What goes with a yes

Second-date decisions are most linked to:

- attractiveness
- fun
- shared interests
- overall liking
- perceived reciprocity

Among the six rated attributes, attractiveness has the strongest link (0.487). In plain numbers: 8% of yes when the partner is rated 4 or less, 76% when rated 7.5 or more.

---

### What people say vs what they do

What participants say they value matches their decisions only in part.

- **Attractiveness ranks first in both**, overall. Women state intelligence first, but their decisions follow attractiveness: a top intelligence rating goes with 46% of yes, a top attractiveness rating with 70%. For men, attractiveness is first in both.
- **Intelligence and sincerity rank lower in decisions** (intelligence: 2nd stated, 4th revealed; sincerity: 4th stated, 5th revealed).
- **Shared interests and fun rank higher in decisions** (shared interests: 5th stated, 3rd revealed; fun: 3rd stated, 2nd revealed).

This is a link, measured with ratings given right after each date. It does not prove that people misjudge their own preferences.

---

### Answers to the questions of the brief

**1. What are the least desirable attributes? Does it differ by gender?**

The attribute least linked to a yes is ambition for a male partner (0.175) and sincerity for a female partner (0.191). The top three are the same for both: attractiveness, fun, shared interests. The bottom values are close, so this difference is small.

**2. How important is attractiveness vs its real impact?**

Overall, attractiveness ranks first in both stated preferences and decisions (correlation 0.487). By gender, women state intelligence first, but their decisions are most linked to attractiveness; for men, attractiveness is first in both.

**3. Are shared interests more important than a shared racial background?**

Yes. Shared interests are much more linked to the decision than same race (0.401 against 0.023). The comparison is not on equal terms: the shared-interests rating is given after the date, while racial background is known before it.

**4. Can people predict their own value in the dating market?**

Not covered in this notebook: it would need the self-ratings and the ratings received, which this analysis does not use.

**5. Is it better to be someone's first date of the night or their last?**

First. The yes rate is 49.9% for first dates against 41.4% for the others: +8.5 points, 95% interval from +4.2 to +12.8. Being last shows no clear effect: +3.4 points, interval from -0.9 to +7.7, which contains 0. The data do not tell us why.

---

### Final takeaway

- **Attractiveness is the strongest of the six rated attributes**, even though women state intelligence first. Overall liking (`like`, 0.513) is a broader measure.
- **Compatibility weighs more in decisions than in statements**: shared interests and fun rank higher in decisions than in stated preferences; intelligence and sincerity lower.

For Tinder, this suggests to:

- put shared interests and compatibility signals forward on profiles
- help users read mutual attraction
- not rely only on photos

These data do not prove that such changes would increase matches. They say what to test, with real users.
